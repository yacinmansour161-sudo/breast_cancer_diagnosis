# Breast Cancer Dataset Exploration

```python
import pandas as pd
import numpy as np
df = pd.read_csv("../data/data.csv")  #load the dataset
df.head() #see the first rows
```

```python
print("shape :", df.shape) #know the shape of the dataset (rows , columns)
print("\ncolumns :\n", df.columns.tolist()) #see the column names (feature names)
```

```python
df = df.drop(columns=["Unnamed: 32"]) #drop the unnamed column
df = df.drop(columns=["id"]) #drop the id column
print(df.shape) #see the new shape
```

```python
df.info() # column types, missing values , memory usage
df.describe() #statistical summary of the dataset(mean,std,min/max,quartiles)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.countplot(x="diagnosis", data=df) #count the number of samples in each class
plt.xlabel("Diagnosis")
plt.ylabel("Count")
plt.title("Data Distribution")
plt.savefig("../figures/data_distribution.png")
plt.show()
```

```python
sns.histplot(data=df, x="radius_mean",hue="diagnosis", kde=True) #histogram of radius_mean feature colored by diagnosis
plt.title("Radius Mean Distribution by Diagnosis")
plt.show()
```

### Observation

The `radius_mean` feature appears to correlate with tumor diagnosis.

- Lower radius values are mostly associated with benign tumors.
- Higher radius values are more associated with malignant tumors.
- However, there is overlap in the middle range, meaning this feature alone cannot perfectly separate the two classes.

This suggests that combining multiple features will be necessary for accurate classification.

```python
df["diagnosis_numeric"] = df["diagnosis"].map({"B": 0, "M": 1})#convert diagnosis into numerical feature
correlation = df.corr(numeric_only=True)#calculate the correlation matrix
plt.figure(figsize=(18,12))#plot the correlation matrix as a heatmap
sns.heatmap(correlation,cmap='coolwarm')#show the heatmap
plt.title("Correlation Heatmap")
plt.savefig("../figures/feature_analysis.png")
plt.show()
```

### Correlation Analysis Observations

The correlation matrix is symmetric because correlation between two variables is mutual.

The diagonal values are equal to 1 since each feature is perfectly correlated with itself.

Several groups of features show strong positive correlations, especially:
- radius-related features,
- perimeter-related features,
- area-related features.

This suggests that some features may contain redundant information.

The target variable (`diagnosis_numeric`) also shows strong correlation with features related to tumor size and concavity, indicating that these variables are important for classification.
